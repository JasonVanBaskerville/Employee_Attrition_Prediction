# 04. Statistical Analysis

## Objective

The objective of this analysis is to formally evaluate relationships and differences identified during exploratory data analysis.

The analysis focuses on:

1. Testing whether categorical employee characteristics are associated with employee attrition.
2. Testing whether numerical employee characteristics differ between employees who experienced attrition and those who remained.
3. Quantifying the uncertainty around the estimated attrition rate.
4. Measuring the magnitude of observed relationships using effect sizes.
5. Translating statistical findings into business-relevant insights without interpreting statistical association as causation.

Important Principle: 

A statistically significant result does not necessarily imply a large or practically important effect. Therefore, p-values will be considered together with confidence intervals and the business context.

## #1 - Categorical vs Categorical

### Case — Overtime vs Attrition

#### Chi-Square Test

Is overtime status related to employee attrition?

H0:
There is no relationship between Overtime and Attrition.

H1:
There is a relationship between Overtime and Attrition.

```python
overtime_table = pd.crosstab(
    df["OverTime"],
    df["Attrition"]
)

overtime_table
```

```python
overtime_rate = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

overtime_rate.round(2)
```

```python
chi2, p_value, dof, expected = chi2_contingency(
    overtime_table
)

print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p_value)
```

Interpretasi

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("There is evidence of the relationship between Overtime and Attrition.")
else:
    print("Fail to reject H0")
    print("There is insufficient evidence of an association.")
```

There is statistical evidence of the relationship between overtime and attrition.

#### Effect Size — Cramér's V

How strong is the relationship?

```python
def cramers_v(table):
    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.to_numpy().sum()
    r, k = table.shape

    phi2 = chi2 / n

    phi2_corrected = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    r_corrected = r - (
        ((r - 1) ** 2) / (n - 1)
    )

    k_corrected = k - (
        ((k - 1) ** 2) / (n - 1)
    )

    denominator = min(
        k_corrected - 1,
        r_corrected - 1
    )

    return np.sqrt(
        phi2_corrected / denominator
    )
```

0.00 → none / very weak
0.10 → small
0.30 → moderate
0.50 → large

```python
cv = cramers_v(overtime_table)

print("Cramér's V:", round(cv, 4))
```

A statistically significant relationship between overtime and attrition was detected, but the effect size was relatively small.

### Other Case

```python
def cramers_v(table):
    """
    Calculate Cramér's V effect size.
    """
    
    chi2, p_value, dof, expected = chi2_contingency(table)
    
    n = table.to_numpy().sum()
    r, k = table.shape
    
    phi2 = chi2 / n
    
    # Bias correction
    phi2_corrected = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )
    
    r_corrected = r - (
        ((r - 1) ** 2) / (n - 1)
    )
    
    k_corrected = k - (
        ((k - 1) ** 2) / (n - 1)
    )
    
    denominator = min(
        k_corrected - 1,
        r_corrected - 1
    )
    
    if denominator == 0:
        return 0
    
    return np.sqrt(
        phi2_corrected / denominator
    )


def interpret_cramers_v(v):
    
    if v < 0.10:
        return "Very Weak"
    
    elif v < 0.30:
        return "Weak"
    
    elif v < 0.50:
        return "Moderate"
    
    else:
        return "Strong"



def chi_square_analysis(
    df,
    column,
    target="Attrition",
    alpha=0.05
):

    data = df[[column, target]].dropna()

    table = pd.crosstab(
        data[column],
        data[target]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        table
    )

    effect_size = cramers_v(table)

    significant = p_value < alpha

    result = {
        "Variable": column,
        "Test": "Chi-Square",
        "Chi2_Statistic": chi2,
        "Degrees_of_Freedom": dof,
        "p_value": p_value,
        "Cramers_V": effect_size,
        "Effect": interpret_cramers_v(effect_size),
        "Significant": significant
    }

    return result
```

```python
categorical_variables = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.drop('Attrition')

categorical_variables = categorical_variables.tolist()

categorical_variables
```

```python
chi_square_results = []

for column in categorical_variables:
    
    result = chi_square_analysis(
        df,
        column,
        target="Attrition"
    )
    
    chi_square_results.append(result)

chi_square_results = pd.DataFrame(
    chi_square_results
)

chi_square_results
```

```python
chi_square_results_sorted = (
    chi_square_results
    .sort_values(
        by=["Significant", "Effect", "Cramers_V"],
        ascending=[False, False, False]
    )
)

chi_square_results_sorted
```

```python
chi_square_results_sorted.loc[chi_square_results_sorted["Significant"] == True, "Variable"].tolist()
```

All categorical features (except Gender and Over18) have a significant relationship with Attrition, with OverTime showing the strongest relationship based on Cramér's V.

Because we do not want to assume normality arbitrarily, use:

#### Mann-Whitney U

H0:
The distribution of YearsAtCompany is the same for both groups.

H1:
The distribution of YearsAtCompany differs between the two groups.

```python
attrition_yes = df.loc[
    df["Attrition"] == "Yes",
    "YearsAtCompany"
]
attrition_yes
```

```python
attrition_no = df.loc[
    df["Attrition"] == "No",
    "YearsAtCompany"
]
attrition_no
```

```python
statistic, p_value = mannwhitneyu(
    attrition_yes,
    attrition_no,
    alternative="two-sided"
)

print("U statistic:", statistic)
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    print(
        "There is statistical evidence that "
        "YearsAtCompany differs between attrition groups."
    )
else:
    print(
        "There is insufficient statistical evidence "
        "of a difference."
    )
```

#### Rank Biserial Correlation

```python
def rank_biserial_correlation(x, y):
    
    u_statistic, _ = mannwhitneyu(
        x,
        y,
        alternative="two-sided"
    )
    
    n1 = len(x)
    n2 = len(y)
    
    rbc = (
        (2 * u_statistic) /
        (n1 * n2)
    ) - 1
    
    return rbc
```

```python
x = df.loc[
    df["Attrition"] == "Yes",
    "YearsAtCompany"
].dropna()

y = df.loc[
    df["Attrition"] == "No",
    "YearsAtCompany"
].dropna()

rbc = rank_biserial_correlation(x, y)

print("Rank-biserial correlation:", round(rbc, 4))
```

|r| Interpretation |
|---:|---|
| < 0.10 | Very small |
| 0.10 - 0.30 | Small |
| 0.30 - 0.50 | Moderate |
| ≥ 0.50 | Large |

### Other Case

```python
def mann_whitney_analysis(
    df,
    column,
    target="Attrition",
    alpha=0.05
):
    
    # =========================
    # Separate groups
    # =========================
    group_yes = df.loc[
        df[target] == "Yes",
        column
    ].dropna()
    
    group_no = df.loc[
        df[target] == "No",
        column
    ].dropna()
    
    
    # =========================
    # Mann-Whitney U Test
    # =========================
    statistic, p_value = mannwhitneyu(
        group_yes,
        group_no,
        alternative="two-sided"
    )
    
    
    # =========================
    # Rank-Biserial Correlation
    # =========================
    n_yes = len(group_yes)
    n_no = len(group_no)
    
    rank_biserial = (
        1 - (2 * statistic) / (n_yes * n_no)
    )
    
    
    # =========================
    # Significance
    # =========================
    significant = p_value < alpha
    
    
    # =========================
    # Result
    # =========================
    result = {
        "Variable": column,
        
        "N_Attrition_Yes": n_yes,
        "N_Attrition_No": n_no,
        
        "Median_Attrition_Yes": group_yes.median(),
        "Median_Attrition_No": group_no.median(),
        
        "Mean_Attrition_Yes": group_yes.mean(),
        "Mean_Attrition_No": group_no.mean(),
        
        "U_Statistic": statistic,
        "p_value": p_value,
        "Significant": significant,
        
        "Rank_Biserial_Correlation": rank_biserial
    }
    
    return result
```

```python
numerical_variables = df.select_dtypes(
    include=["int64", "float64"]
).columns
```

```python
mann_whitney_results = []

for column in numerical_variables:
    result = mann_whitney_analysis(
        df,
        column,
        target="Attrition"
    )
    
    mann_whitney_results.append(result)

mann_whitney_results = pd.DataFrame(mann_whitney_results)

mann_whitney_results
```

```python
mann_whitney_results_sorted = (
    mann_whitney_results
    .assign(
        Effect_Size_Abs=mann_whitney_results[
            "Rank_Biserial_Correlation"
        ].abs()
    )
    .sort_values(
        by=["Significant", "Effect_Size_Abs"],
        ascending=[False, False]
    )
    .drop(columns="Effect_Size_Abs")
)

mann_whitney_results_sorted
```

TotalWorkingYears memiliki distribusi yang berbeda secara signifikan antara karyawan yang mengalami attrition dan yang tidak, dengan rank-biserial correlation sebesar 0.31.

```python
mann_whitney_results_sorted.loc[mann_whitney_results_sorted["Significant"] == True, "Variable"].tolist()
```

## #3 Confidence Interval for Attrition Rate

```python
attrition_count = (
    df["Attrition"] == "Yes"
).sum()

total_employees = len(df)

attrition_rate = (
    attrition_count / total_employees
)

print(
    f"Attrition rate: "
    f"{attrition_rate * 100:.2f}%"
)
```

How uncertain is that estimate?

```python
lower, upper = proportion_confint(
    count=attrition_count,
    nobs=total_employees,
    alpha=0.05,
    method="wilson"
)

print(
    f"Attrition rate: {attrition_rate * 100:.2f}%"
)

print(
    f"95% CI: "
    f"{lower * 100:.2f}% - "
    f"{upper * 100:.2f}%"
)
```

Estimated attrition rate is 16.12%, with a 95% confidence interval of approximately 14.30% – 18.05%.

### Confidence Interval based on Department (if need)

```python
def attrition_rate_ci(df, group_column):
    
    results = []
    
    for group, data in df.groupby(group_column):
        
        n = len(data)
        
        count = (
            data["Attrition"] == "Yes"
        ).sum()
        
        rate = count / n
        
        lower, upper = proportion_confint(
            count=count,
            nobs=n,
            alpha=0.05,
            method="wilson"
        )
        
        results.append({
            group_column: group,
            "n": n,
            "attrition_count": count,
            "attrition_rate": rate,
            "ci_lower": lower,
            "ci_upper": upper
        })
    
    return pd.DataFrame(results)
```

```python
department_ci = attrition_rate_ci(
    df,
    "Department"
)

department_ci
```

```python
department_ci[
    [
        "attrition_rate",
        "ci_lower",
        "ci_upper"
    ]
] *= 100

department_ci
```

#### Visualisasi Confidence Interval

```python
plt.figure(figsize=(10, 6))

plt.errorbar(
    department_ci["attrition_rate"],
    department_ci["Department"],
    xerr=[
        department_ci["attrition_rate"]
        - department_ci["ci_lower"],
        
        department_ci["ci_upper"]
        - department_ci["attrition_rate"]
    ],
    fmt="o"
)

plt.xlabel("Attrition Rate (%)")
plt.ylabel("Department")
plt.title(
    "Attrition Rate by Department "
    "with 95% Confidence Intervals"
)

plt.show()
```
