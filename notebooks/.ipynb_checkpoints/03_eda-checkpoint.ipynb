# 03 Exploratory Data Analysis

## Objective

The objective of this exploratory data analysis is to identify patterns
and relationships between employee characteristics and employee attrition.

The analysis focuses on the following business questions:
1. What is the overall employee attrition rate?
2. Which departments and job roles have higher attrition rates?
3. Is attrition associated with overtime?
4. How does job satisfaction relate to attrition?
5. Are employees with shorter tenure more likely to leave?
6. Are there differences in income, age, and working experience between employees who leave and those who remain?
7. Which variables appear to be potentially associated with attrition and should be investigated further using statistical testing?

## Import Library

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)

sns.set_theme(style="whitegrid")
```

## Load Dataset

```python
DATA_PATH = Path("../data/clean/clean_employee_attrition.csv")

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
df.shape
```

## EDA #1 — Overall Attrition

What is the overall employee attrition rate?

```python
attrition_count = df["Attrition"].value_counts()

attrition_percentage = (
    df["Attrition"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(attrition_count)
print()
print(attrition_percentage)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Attrition"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

## EDA #2 — Job

### Job Level

How does job level relate to attrition?

```python
job_level_attrition = (
    df.groupby("JobLevel")["Attrition"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

job_level_attrition
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=job_level_attrition.index,
    y=job_level_attrition.values
)

plt.xticks(
    ticks=[0, 1, 2, 3, 4], 
    labels=["Entry-Level", "Mid-Level", "Supervisory", "Managerial", "Executive"]
)

plt.title("Attrition Rate by Job Level")
plt.xlabel("Job Level")
plt.ylabel("Attrition Rate (%)")

plt.show()
```

## EDA #17 — Correlation Between Numerical Features vs Target

Which numerical features show a high correlation with the target?

```python
df["Attrition_binary"] = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})
```

```python
numeric_features = df.select_dtypes(
    include=["int64", "float64"]
)

correlation_target = (
    numeric_features
    .corr()["Attrition_binary"]
    .sort_values()
)
```

```python
plt.figure(figsize=(8, 10))

correlation_target.drop(
    "Attrition_binary"
).plot(
    kind="barh"
)

plt.title("Correlation of Numerical Features with Attrition")
plt.xlabel("Correlation")

plt.show()
```

## EDA #18 — Employee Profile

```python
profile = (
    df.groupby("Attrition")
      .agg(
          Average_Age=("Age", "mean"),
          Average_Income=("MonthlyIncome", "mean"),
          Average_Tenure=("YearsAtCompany", "mean"),
          Average_Working_Years=("TotalWorkingYears", "mean"),
          Average_Distance=("DistanceFromHome", "mean")
      )
      .round(2)
)

profile
```

## EDA #19 — Summary of Attrition

```python
def attrition_rate_by_column(df, column):
    result = (
        df.groupby(column)["Attrition"]
          .apply(lambda x: (x == "Yes").mean() * 100)
          .sort_values(ascending=False)
    )
    
    return result
```

```python
attrition_rate_by_column(df, "Department")
```

## Key EDA Findings

Based on the exploratory analysis, several patterns were identified:

1. Overall Attrition

The dataset shows an imbalance between employees who remained and
employees who experienced attrition.

This indicates that model evaluation should not rely solely on accuracy.

2. Department and Job Role

Attrition rates vary across departments and job roles. Some employee
groups show relatively higher attrition rates and may require further
investigation.

3. Overtime

Employees with different overtime statuses show different attrition
rates. This relationship should be formally evaluated using an
appropriate statistical test.

4. Job Satisfaction

Attrition rates vary across job satisfaction levels. This suggests that
job satisfaction may be associated with employee attrition, although
the EDA alone cannot establish causality.

5. Tenure

Employees with different lengths of service show different attrition
patterns. Tenure-related variables should therefore be investigated
further.

6. Numerical Variables

Variables such as age, monthly income, distance from home, and years at
the company show differences between employees who left and those who
remained.

These differences should be evaluated statistically before drawing
strong conclusions.

Next Step

The next stage is statistical analysis to determine whether the
observed differences and associations are statistically significant
and to quantify their uncertainty.

```python
# | Observasi EDA                         | Statistical Question                       | Test yang mungkin     |
# | ------------------------------------- | ------------------------------------------ | --------------------- |
# | Overtime punya attrition rate berbeda | Apakah Overtime dan Attrition berhubungan? | Chi-Square            |
# | Job Role berbeda attrition rate       | Apakah Job Role dan Attrition berhubungan? | Chi-Square            |
# | Age berbeda antara Yes/No             | Apakah distribusi Age berbeda?             | Mann-Whitney / t-test |
# | Monthly Income berbeda                | Apakah income berbeda antar grup?          | Mann-Whitney / t-test |
# | YearsAtCompany berbeda                | Apakah tenure berbeda antar grup?          | Mann-Whitney / t-test |
# | DistanceFromHome berbeda              | Apakah jarak berbeda antar grup?           | Mann-Whitney / t-test |
```
