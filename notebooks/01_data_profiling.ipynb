# Data Profilling

## Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

sns.set_theme(style="whitegrid")
```

## Load Dataset

```python
DATA_PATH = Path("../data/raw/employee_attrition.csv")
```

```python
DATA_PATH.parent
```

```python
df = pd.read_csv(DATA_PATH)
```

## Dataset Overview

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.head()
```

```python
df.tail()
```

## Dataset Information

```python
df.info()
```

```python
numeric_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns

print("Numerical columns:")
print(list(numeric_columns))

print("\nCategorical columns:")
print(list(categorical_columns))
```

## Missing Value

```python
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (
        df.isnull().mean() * 100
    )
})

missing = missing.sort_values(
    "missing_percentage",
    ascending=False
)

missing
```

```python
missing_plot = missing[
    missing["missing_count"] > 0
]

plt.figure(figsize=(10, 5))

sns.barplot(
    data=missing_plot.reset_index(),
    x="missing_percentage",
    y="index"
)

plt.xlabel("Missing Values (%)")
plt.ylabel("Column")
plt.title("Missing Value Analysis")

plt.show()
```

No missing values were identified in the dataset.

## Duplicate Records

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
duplicate_percentage = (
    duplicate_count / len(df) * 100
)

print(
    f"Duplicate percentage: "
    f"{duplicate_percentage:.2f}%"
)
```

No duplicate row were identified in the dataset.

## Unique Values

```python
unique_summary = pd.DataFrame({
    "column": df.columns,
    "unique_values": [
        df[col].nunique()
        for col in df.columns
    ]
})

unique_summary.sort_values(
    "unique_values"
)
```

```python
for col in categorical_columns:
    print(f"\n===== {col} =====")
    print(df[col].value_counts(dropna=False).head(10))
```

## Numerical Summary

```python
numeric_summary = df[numeric_columns].describe().T

numeric_summary["missing"] = (
    df[numeric_columns].isnull().sum()
)

numeric_summary["unique"] = (
    df[numeric_columns].nunique()
)

numeric_summary
```

## Categorical Summary

```python
categorical_summary = pd.DataFrame({
    "column": categorical_columns,
    "unique_values": [
        df[col].nunique()
        for col in categorical_columns
    ],
    "missing": [
        df[col].isnull().sum()
        for col in categorical_columns
    ]
})

categorical_summary
```

## Target Variable

```python
target = "Attrition"
df[target].value_counts()
```

```python
df[target].value_counts(normalize=True) * 100
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.show()
```

```python
attrition_rate = (
    df[target].eq("Yes").mean() * 100
)

print(
    f"Overall attrition rate: "
    f"{attrition_rate:.2f}%"
)
```

Approximately 16.12% of employees left, while about 83.88% did not. This situation indicates a class imbalance that needs to be addressed when building a classification model.

## Potential ID Columns

```python
Potential_ID_columns = []
for col in df.columns:
    if df[col].nunique() == len(df):
        Potential_ID_columns.append(col)
        print(
            f"{col}: potentially unique identifier"
        )
```

```python
Potential_ID_columns
```

## Basic Data Quality Check

```python
data_quality = pd.DataFrame({
    "Issue": [
        "Missing values",
        "Duplicate rows",
        "Potential ID columns",
        "Target imbalance"
    ],
    "Finding": [
        f"{df.isnull().sum().sum()} missing values",
        f"{df.duplicated().sum()} duplicate rows",
        f"{', '.join(Potential_ID_columns)}",
        f"Attrition rate = {attrition_rate:.2f}%"
    ]
})

data_quality
```
