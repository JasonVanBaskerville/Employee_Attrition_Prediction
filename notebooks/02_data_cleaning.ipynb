# 02. Data Cleaning

## Objective

The data cleaning stage aims to ensure the Employee Attrition dataset is of high quality before being used in the Exploratory Data Analysis (EDA) and Statistical Analysis stages.

The cleaning process includes:

1. Checking the dataset structure and quality.
2. Identifying duplicate data.
3. Identifying missing values.
4. Checking for data type inconsistencies.
5. Identifying invalid or inconsistent values.
6. Handling duplicate data and missing values if found.
7. Validating the Attrition target variable.
8. Saving the cleaned dataset.

The output of this notebook is a cleaned dataset that will serve as input for the next stage.

## Duplicate Data

```python
data = data.drop_duplicates()
```

## Missing Values

```python
missing_values = (
    data.isnull()
    .sum()
    .sort_values(ascending=False)
)

missing_values
```

Fill in categorical data with the mode.

```python
numeric_missing = [
    col for col in numeric_columns
    if data[col].isnull().sum() > 0
]

for column in numeric_missing:
    data[column] = data[column].fillna(
        data[column].median()
    )
numeric_missing
```

```python
categorical_missing = [
    col for col in categorical_columns
    if data[col].isnull().sum() > 0
]

for column in categorical_missing:
    data[column] = data[column].fillna(
        data[column].mode()[0]
    )
categorical_missing
```

## Data Types

```python
dtype = pd.DataFrame({
    "dtype": data.dtypes
})

dtype
```

```python
categorical_columns = data.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numeric_columns = data.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical columns:")
print(categorical_columns)

print("\nNumeric columns:")
print(numeric_columns)
```

## Unique Values

## Invalid Values

```python
data[numeric_columns].describe().T
```

```python
negative_values = {}

for column in numeric_columns:
    count = (data[column] < 0).sum()
    
    if count > 0:
        negative_values[column] = count

negative_values
```

There are no negative values in the numerical features.

## Zero Values

```python
zero_summary = pd.DataFrame({
    "zero_count": (data[numeric_columns] == 0).sum(),
})

zero_summary = zero_summary.sort_values(
    "zero_count",
    ascending=False
)

zero_summary
```

## Inconsistent Values

```python
for column in categorical_columns:
    data[column] = data[column].astype(str).str.strip()
```

```python
for column in categorical_columns:
    print(f"\n{column}")
    print(data[column].unique())
```

## Target Value Validation

```python
TARGET = "Attrition"

print(data[TARGET].value_counts(dropna=False))
```

```python
attrition_distribution = pd.DataFrame({
    "count": data[TARGET].value_counts(),
    "percentage": data[TARGET].value_counts(normalize=True) * 100
})

attrition_distribution
```

```python
valid_attrition = {"Yes", "No"}

invalid_attrition = set(data[TARGET].dropna().unique()) - valid_attrition

print("Invalid Attrition values:")
print(invalid_attrition)
```

## Check ID / Identifier Columns

```python
id_like_columns = [
    column for column in data.columns
    if "id" in column.lower()
    or "number" in column.lower()
]

print(id_like_columns)
```

## Final Quality Report

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Duplicate Rows",
        "Missing Values",
        "Numeric Features",
        "Categorical Features",
        "Target Variable"
    ],
    
    "Value": [
        data.shape[0],
        data.shape[1],
        data.duplicated().sum(),
        data.isnull().sum().sum(),
        len(numeric_columns),
        len(categorical_columns),
        TARGET
    ]
})

quality_report
```

## Save Clean Dataset

```python
OUTPUT_PATH = Path(
    "../data/clean/clean_employee_attrition.csv"
)

OUTPUT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

data.to_csv(
    OUTPUT_PATH,
    index=False
)

print(f"Clean dataset saved to: {OUTPUT_PATH}")
```
